# 27l Soft-Leak Brightness-Routed MoE

27k showed the first clear gain by routing day-like and night-like images to separate specialist model pools. This run keeps that hard route but leaks low-score boxes from the opposite pool to recover ambiguous images without fully reintroducing noisy experts.

```python
from pathlib import Path
import subprocess
import sys

repo = Path('/app/Object_Detection')
script = repo / 'dynamic_quality_aware_classwise_aggregation/scene_daynight_dqa/aggressive_dqamox/scripts/build_eval_27l_soft_leak_routed_moe.py'
workspace = repo / 'dynamic_quality_aware_classwise_aggregation/scene_daynight_dqa/aggressive_dqamox/output/27l_soft_leak_routed_moe_r1'
cmd = [sys.executable, str(script), '--target-map50', '0.60', '--workspace', str(workspace)]
print('Running:', ' '.join(cmd))
result = subprocess.run(cmd, cwd=repo)
if result.returncode not in (0, 2):
    raise SystemExit(result.returncode)
print('27l completed with returncode', result.returncode)
```

# 27f Post-hoc Expert Graft

27b-27e は warmup を少しだけ超えるが、平均 aggregation で local expert が薄まっている。FedDG-MoE の test-time adapter fusion と、pseudo-label quality/long-tail weighting 系の示唆を使い、学習を増やす前に client MoE residual を単一 checkpoint の expert slot に詰め替えて total mAP50 を評価する。

```python
import subprocess, sys
from pathlib import Path

ROOT = Path('/app/Object_Detection')
SCRIPT = ROOT / 'dynamic_quality_aware_classwise_aggregation' / 'scene_daynight_dqa' / 'aggressive_dqamox' / 'scripts' / 'build_eval_27f_posthoc_expert_graft.py'
cmd = [sys.executable, str(SCRIPT), '--target-map50', '0.60', '--splits', 'total', '--batch-size', '16']
print(' '.join(cmd))
subprocess.run(cmd, cwd=ROOT, check=True)
```

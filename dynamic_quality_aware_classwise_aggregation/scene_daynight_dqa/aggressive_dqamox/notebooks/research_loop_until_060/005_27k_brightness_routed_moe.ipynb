# 27k Brightness-Routed Model-Level MoE

- target: scene_daynight_total mAP50 >= 0.600
- previous best: 27h/27i/27j around mAP50=0.464
- idea: keep expert models separate at inference and route by image brightness, instead of pushing pseudo-label gradients into one averaged checkpoint.

## Paper Basis

- [HI-MoE](https://arxiv.org/abs/2604.04908): object detection MoE should route hierarchically by scene and instance, not only with one image-level expert.
- [Domain-Specialized Object Detection via Model-Level Mixtures of Experts](https://arxiv.org/abs/2604.18256): BDD100K detection can benefit from model-level expert fusion and domain-specialized expert pools.
- [SEAS-GMoE](https://www.sciencedirect.com/science/article/pii/S0957417426000424): decoupling routing from gradient noise motivates a test-time router after 27j showed pseudo-label distillation does not move total mAP.


```python
import subprocess
import sys
from pathlib import Path

REPO_ROOT = Path('/app/Object_Detection')
SCRIPT = REPO_ROOT / 'dynamic_quality_aware_classwise_aggregation/scene_daynight_dqa/aggressive_dqamox/scripts/build_eval_27k_brightness_routed_moe.py'
WORKSPACE = REPO_ROOT / 'dynamic_quality_aware_classwise_aggregation/scene_daynight_dqa/aggressive_dqamox/output/27k_brightness_routed_moe_r2'
CMD = [sys.executable, str(SCRIPT), '--target-map50', '0.60', '--workspace', str(WORKSPACE)]
print(' '.join(CMD))
proc = subprocess.run(CMD, cwd=REPO_ROOT, check=False)
print('returncode', proc.returncode)
if proc.returncode not in (0, 2):
    raise SystemExit(proc.returncode)
```

27s showed that tiled inference helps only slightly. This notebook tests a more aggressive MoE router: use the known scene/day-night split from the image path to choose domain-specific experts, then gate before any full run.

```python
from pathlib import Path
import subprocess

repo = Path('/app/Object_Detection')
script = repo / 'dynamic_quality_aware_classwise_aggregation/scene_daynight_dqa/aggressive_dqamox/scripts/build_eval_27t_path_domain_routed_moe.py'
workspace = repo / 'dynamic_quality_aware_classwise_aggregation/scene_daynight_dqa/aggressive_dqamox/output/27t_path_domain_routed_moe_r1'
cmd = [
    '/root/micromamba/envs/al_yolov8/bin/python',
    str(script),
    '--workspace', str(workspace),
    '--target-map50', '0.60',
    '--previous-best-map50', '0.52939',
    '--gate-images', '1200',
    '--min-gate-gain', '0.010',
]
print(' '.join(cmd))
result = subprocess.run(cmd, cwd=repo)
print('returncode:', result.returncode)
```
